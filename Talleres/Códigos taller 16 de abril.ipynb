```python
#Punto 1

# Mezcla de fertilizantes

import numpy as np
from scipy.optimize import linprog

# Coeficientes de la función objetivo
# linprog MINIMIZA, así que ponemos negativos para maximizar
c = [-180, -220]

# Matriz de restricciones (lado izquierdo)
A = [
    [3, 4],   # Compuesto A
    [2, 1],   # Compuesto B
    [5, 3]    # Compuesto C
]

# Lado derecho
b = [120, 80, 150]

# Límites de variables (x1 >= 0, x2 >= 0)
bounds = [(0, None), (0, None)]

# Resolver
resultado = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method='highs')

# Resultados
x1_opt, x2_opt = resultado.x
Z_opt = -resultado.fun

print()
print("Mezcla de fertilizantes (PL)")
print()

print(f"Estado: {'Óptimo encontrado' if resultado.success else 'Fallo'}")
print()
print(f"x1 (F1) = {x1_opt:.2f} ton")
print(f"x2 (F2) = {x2_opt:.2f} ton")
print(f"Ganancia máxima = ${Z_opt:.2f}")
print()
```

```python
#Punto 2

import numpy as np

# Parámetros iniciales
S0 = 200
M0 = 80

# Número de semanas a simular
n = 2

# Listas para guardar resultados
S = [S0]
M = [M0]

# Iteración
for i in range(n):
    S_next = 0.6*S[i] + 0.2*M[i] + 40
    M_next = 0.1*S[i] + 0.5*M[i] + 20

    S.append(S_next)
    M.append(M_next)

# Resultados
print("Producción semanal")
print("="*40)

for i in range(len(S)):
    print(f"Semana {i}: Sillas = {S[i]:.2f}, Mesas = {M[i]:.2f}")
```

```python
#Punto 3

import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Parámetros
beta = 0.00003
gamma = 0.1
N = 10000

# Modelo
def modelo(t, I):
    return beta * (N - I) * I - gamma * I

# Condición inicial
I0 = [50]

# Tiempo (días)
t_span = (0, 100)
t_eval = np.linspace(0, 100, 200)

# Resolver
sol = solve_ivp(modelo, t_span, I0, t_eval=t_eval)

# Valor final aproximado
print(f"Infectados al final: {sol.y[0][-1]:.2f}")
```

```python
#Punto 4

import numpy as np

# Parámetros
r = 0.4
K = 1000
H = 80

# Condición inicial
P0 = 400

# Número de años
n = 10

# Lista
P = [P0]

# Iteración
for i in range(n):
    P_next = P[i] + r*P[i]*(1 - P[i]/K) - H
    P.append(P_next)

# Resultados
print("Modelo logístico con pesca")
print()

for i in range(len(P)):
    print(f"Año {i}: Población = {P[i]:.2f}")
```
